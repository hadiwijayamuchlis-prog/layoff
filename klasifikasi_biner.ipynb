# Klasifikasi Biner: Pima Indians Diabetes

**Sumber dataset:** OpenML (dataset `diabetes`, versi 1), berasal dari National Institute of Diabetes and Digestive and Kidney Diseases (tersedia juga di UCI/Kaggle). Dimuat lewat `sklearn.datasets.fetch_openml`.

**Jenis data:** tabular numerik (data medis pasien perempuan suku Pima, usia minimal 21 tahun), bukan teks atau image.

> Butuh koneksi internet saat pertama kali dijalankan untuk mengunduh dataset.

```python
import pandas as pd
from sklearn.datasets import fetch_openml

raw = fetch_openml("diabetes", version=1, as_frame=True, parser="auto")
df = raw.frame       # DataFrame berisi fitur + kolom 'class'
print(df.shape)
df.head()
```

## Identifikasi fitur dan target

```python
target_col = "class"
X = df.drop(columns=target_col)   # fitur
y = df[target_col]                # target / label

print("Jumlah data    :", df.shape[0])
print("Jumlah fitur   :", X.shape[1])
print("Jumlah target  :", 1)
print("Total variabel :", df.shape[1], "(fitur + target)")
print("\nNama fitur :", list(X.columns))
print("Nama target:", target_col)
```

Keterangan fitur: `preg` (jumlah kehamilan), `plas` (kadar glukosa plasma), `pres` (tekanan darah diastolik), `skin` (tebal lipatan kulit), `insu` (insulin serum), `mass` (BMI), `pedi` (riwayat diabetes keluarga), `age` (usia).

```python
print("Distribusi kelas:")
print(y.value_counts())
print("\nJumlah kelas unik:", y.nunique())
print("Nilai kelas      :", list(y.unique()))
print("Missing value    :", df.isna().sum().sum())
```

## Kesimpulan

```python
n = y.nunique()
if n == 2:
    jenis = "Klasifikasi Biner"
elif n > 2:
    jenis = "Klasifikasi Multikelas"
print("Jenis klasifikasi:", jenis)
```

**Kesimpulan:** dataset berisi 768 data, 8 fitur, dan 1 kolom target. Target hanya punya 2 kelas (`tested_positive` dan `tested_negative`), jadi ini **Klasifikasi Biner**. Data berupa tabular numerik, bukan teks atau image.
