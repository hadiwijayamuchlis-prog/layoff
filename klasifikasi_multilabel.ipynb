# Klasifikasi Multilabel: Yeast (Fungsi Gen)

**Sumber dataset:** OpenML, dataset `yeast` (versi 4, multilabel), dimuat lewat `sklearn.datasets.fetch_openml`. Dataset asli dari Elisseeff & Weston (2002).

**Jenis data:** tabular numerik (data ekspresi gen mikroarray), bukan teks atau image.

> Butuh koneksi internet saat pertama kali dijalankan untuk mengunduh dataset.

```python
import pandas as pd
from sklearn.datasets import fetch_openml

raw = fetch_openml("yeast", version=4, as_frame=True, parser="auto")
df = raw.frame
print(df.shape)
df.head()
```

## Identifikasi fitur dan label

```python
# Kolom label biasanya bernama Class1..Class14, fitur bernama Att1..Att103.
# Cek nama kolom bila berbeda:
print(list(df.columns[:5]), "...", list(df.columns[-16:]))
```

```python
label_cols = [c for c in df.columns if c.lower().startswith("class")]
feature_cols = [c for c in df.columns if c not in label_cols]

X = df[feature_cols].astype(float)
# Label di OpenML berupa string 'TRUE'/'FALSE' (atau 0/1) -> ubah ke 0/1
y = df[label_cols].astype(str).apply(lambda s: s.str.lower().isin(["true", "1"])).astype(int)

print("Jumlah data    :", df.shape[0])
print("Jumlah fitur   :", X.shape[1])
print("Jumlah label   :", y.shape[1], "kolom target")
print("Total variabel :", df.shape[1])
print("\nKolom label:", label_cols)
```

```python
print("Distribusi per label (jumlah data bernilai 1):")
print(y.sum())

label_per_data = y.sum(axis=1)
print("\nRata-rata label per data:", round(label_per_data.mean(), 2))
print("Maksimum label per data :", label_per_data.max())
print("Missing value           :", df.isna().sum().sum())
```

## Kesimpulan

```python
if y.shape[1] == 1 and y.iloc[:, 0].nunique() == 2:
    jenis = "Klasifikasi Biner"
elif y.shape[1] == 1:
    jenis = "Klasifikasi Multikelas"
elif label_per_data.max() > 1:
    jenis = "Klasifikasi Multilabel"
else:
    jenis = "Klasifikasi Multikelas (one-hot)"
print("Jenis klasifikasi:", jenis)
```

**Kesimpulan:** dataset berisi sekitar 2.417 data, 103 fitur, dan 14 kolom label biner. Satu gen bisa memiliki lebih dari satu label fungsi sekaligus (rata-rata lebih dari 1 label per data), jadi ini **Klasifikasi Multilabel**. Angka pastinya akan tampil setelah sel di atas dijalankan.
