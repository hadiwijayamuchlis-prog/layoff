# Klasifikasi Multikelas: Digits (Angka Tulisan Tangan)

**Sumber dataset:** UCI ML Repository (Optical Recognition of Handwritten Digits), dimuat lewat `sklearn.datasets.load_digits`.

**Jenis data:** **image**. Setiap gambar berukuran 8x8 piksel, diratakan menjadi 64 fitur.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

data = load_digits(as_frame=True)
df = data.frame          # 64 kolom piksel + kolom 'target'
df.head()
```

## Identifikasi fitur dan target

```python
X = df.drop(columns="target")   # fitur: 64 nilai piksel
y = df["target"]                # target: angka 0-9

print("Jumlah data    :", df.shape[0])
print("Jumlah fitur   :", X.shape[1], "(8x8 piksel)")
print("Jumlah target  :", 1)
print("Total variabel :", df.shape[1], "(fitur + target)")
print("\nKelas target   :", sorted(y.unique()))
```

```python
print("Distribusi kelas:")
print(y.value_counts().sort_index())
print("\nJumlah kelas unik:", y.nunique())
print("Missing value    :", df.isna().sum().sum())
```

```python
# Contoh gambar untuk membuktikan data berupa image
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for ax, img, label in zip(axes.ravel(), data.images[:10], y[:10]):
    ax.imshow(img, cmap="gray_r")
    ax.set_title(f"Label: {label}")
    ax.axis("off")
plt.tight_layout()
plt.show()
```

## Kesimpulan

```python
n = y.nunique()
if n == 2:
    jenis = "Klasifikasi Biner"
elif n > 2:
    jenis = "Klasifikasi Multikelas"
print("Jenis klasifikasi:", jenis)
```

**Kesimpulan:** dataset berisi 1.797 data, 64 fitur (piksel), dan 1 kolom target. Target punya 10 kelas (angka 0-9) dan setiap gambar hanya memiliki satu kelas, jadi ini **Klasifikasi Multikelas**. Jenis datanya **image**.
